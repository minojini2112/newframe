```python
# In Jupyter: use `!` for shell commands, not `%` (magics like %timeit).
# AWS CLI is optional — boto3 works anonymously for public NOAA buckets.

%pip install -q boto3

import boto3
from botocore import UNSIGNED
from botocore.config import Config

s3 = boto3.client("s3", config=Config(signature_version=UNSIGNED))
# Hour 12 UTC had no files yet; hour 05 has C13 (M6C13) scans for DOY 181.
prefix = "ABI-L1b-RadF/2026/181/05/"

paginator = s3.get_paginator("list_objects_v2")
keys = []
for page in paginator.paginate(Bucket="noaa-goes19", Prefix=prefix):
    for obj in page.get("Contents", []):
        if "C13" in obj["Key"]:
            keys.append(obj["Key"])

print(f"Found {len(keys)} C13 files under s3://noaa-goes19/{prefix}")
for key in keys[:20]:
    print(key)
if len(keys) > 20:
    print(f"... and {len(keys) - 20} more")
```
